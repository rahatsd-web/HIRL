Library call

```python
import cv2
import matplotlib.pyplot as plt
import numpy as np
```

```python
image = cv2.imread(image_path)

print(image.shape)
```

Import original image

```python
image_rgb = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)

plt.figure(figsize=(6,6))
plt.imshow(image_rgb)
plt.title("Original MRI Image")
plt.axis("off")
plt.show()
```

Converting RGB - Grayscale

```python
gray_image = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)


plt.figure(figsize=(6,6))
plt.imshow(gray_image, cmap="gray")
plt.title("Grayscale MRI Image")
plt.axis("off")
plt.show()
```

Enhanced image

```python
hist_eq = cv2.equalizeHist(gray_image)


plt.figure(figsize=(6,6))
plt.imshow(hist_eq, cmap="gray")
plt.title("Histogram Equalization")
plt.axis("off")
plt.show()
```

```python
clahe = cv2.createCLAHE(clipLimit=2.0,tileGridSize=(8,8))

clahe_image = clahe.apply(gray_image)

plt.figure(figsize=(6,6))
plt.imshow(clahe_image, cmap="gray")
plt.title("CLAHE Enhanced Image")
plt.axis("off")
plt.show()
```

```python
median_filtered = cv2.medianBlur(image_rgb,5)

plt.figure(figsize=(6,6))

plt.imshow(median_filtered, cmap="gray")
plt.title("Median Filtered MRI Image")
plt.axis("off")

plt.show()
```

```python
nlm_image = cv2.fastNlMeansDenoising(image_rgb, None, h=10, templateWindowSize=7, searchWindowSize=21)

plt.figure(figsize=(6,6))
plt.imshow(nlm_image)
plt.title("NlMeans Denoised MRI Image")
plt.axis("off")
plt.show()
```

Enhanced image outputs together (Original + Median Filter + NLM + Histogram Equalization + CLAHE)

```python
plt.figure(figsize=(20,8))


# Original
plt.subplot(2,3,1)
plt.imshow(gray_image, cmap="gray")
plt.title("Original MRI")
plt.axis("off")


# Median Filter
plt.subplot(2,3,2)
plt.imshow(median_filtered, cmap="gray")
plt.title("Median Filtering")
plt.axis("off")


# NLM
plt.subplot(2,3,3)
plt.imshow(nlm_image, cmap="gray")
plt.title("NLM Filtering")
plt.axis("off")


# NLM
plt.subplot(2,3,3)
plt.imshow(gray_image, cmap="gray")
plt.title("NLM Filtering")
plt.axis("off")


# Histogram Equalization
plt.subplot(2,3,4)
plt.imshow(hist_eq, cmap="gray")
plt.title("Histogram Equalization")
plt.axis("off")


# CLAHE
plt.subplot(2,3,5)
plt.imshow(clahe_image, cmap="gray")
plt.title("CLAHE Enhancement")
plt.axis("off")


plt.tight_layout()
plt.show()
```
